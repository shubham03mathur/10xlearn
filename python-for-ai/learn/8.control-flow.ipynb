# Control Flow

Control flow refers to the order == in which the code or function is executed. We use control flow to decide which parts of the code run based on conditions or repetitions.

### Conditionals (if/elif/else)
Conditionals allow us to run specific code only if a condition is met.

```python
age = 18

if age < 13:
    print("You are a child.")
elif age < 20:
    print("You are a teenager.")
else:
    print("You are an adult.")
```

### Loops (for/while)

Loops are used to repeat a block of code multiple times.

**For loops** are typically used when you know the number of iterations (e.g., iterating over a list).
**While loops** are used when you want to repeat until a condition becomes false.

```python
# For loop example
fruits = ["Apple", "Banana", "Cherry"]
for fruit in fruits:
    print(f"I love {fruit}")

# While loop example
count = 1
while count <= 3:
    print(f"Counting: {count}")
    count += 1
```
