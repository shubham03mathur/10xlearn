# File Handling

Handling files is a core skill in Python. Whether you are reading datasets, saving model weights, or logging errors, you need to know how to work with files safely.

### The `with` Statement

The best practice for opening files in Python is using the `with` statement. It ensures that the file is automatically closed after the operations are finished, even if an error occurs.

```python
# Writing to a file
with open("example.txt", "w") as f:
    f.write("Hello, this is a test file.")

print("File written successfully.")

# Reading from a file
with open("example.txt", "r") as f:
    content = f.read()
    print(f"Content: {content}")
```

### Working with JSON

In machine learning, we often save configurations or data as JSON files. Python's `json` module makes this very easy.

```python
import json

data = {
    "model": "RandomForest",
    "hyperparameters": {
        "n_estimators": 100,
        "max_depth": 10
    }
}

# Save to JSON
with open("config.json", "w") as f:
    json.dump(data, f, indent=4)

# Load from JSON
with open("config.json", "r") as f:
    loaded_data = json.load(f)

print(loaded_data["model"])
```
