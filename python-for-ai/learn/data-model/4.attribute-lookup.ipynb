## TL;DR
Customizing attribute access (__getattribute__, __getattr__, __setattr__, __delattr__) and the descriptor mechanism. This is one of the most fundamental—and most frequently misunderstood—parts of Python's object model.

Consider this simple line:

```
person.name

```

It looks trivial but internally python performs sophesticated internal algorithm to find (or try to find) the value.
How does python knows where is `name` property:  is on instance? or class? or a parent class? or built-ins?

the answer is:
> Python fdllows a precision lookup algorithm.

Where can attribute live?

1. Instance Level:

````python
class Person():
    def __init__(self, name):
        self.name = name

p = Person("Shubham")
p.name
````

2. Class Level (class attribute):

````python
class Person():
    source = "github"
    def __init__(self, name):
            self.name = name

Person.source
````

This also works:

````python
p = Person("shubham")
p.source
````

How did it work? Even when the source attribute was never asssigned to instance?

> Because Python searches the class if it doesn't find the attribute on the instance.

### Instance vs Class Attributes
```
           Person Class
        ┌─────────────────┐
        │ species="Human" │
        └────────▲────────┘
                 │
        ┌────────┴────────┐
        │                 │
   Person("Alice")   Person("Bob")
        │                 │
   name="Alice"      name="Bob"

```

The class attribute is shared across all the instances whereas the instance level attributes are instance level (obviously!). This distinction matters as instance level attributes are essentially in-memoryand are unique per instance.
