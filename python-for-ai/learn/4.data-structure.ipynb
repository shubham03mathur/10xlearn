# List

Lists are mutable sequences, typically used to store collections of homogeneous items but not necessarily, In fact Python List can surely store different type of data. For ex:

````python
studentA = ["shubham", 29, '5\'7"', 'Maths']
for student_meta in studentA:
    print(student_meta)
````

## Here are some quirks about the Python lists (Take a note for an interview)

### 1. Is a Python list stored in contiguous (sequential) memory?

Partly.

Python stores the memory pointer in contigous memory location, not the actual value, since everything in Python is a Object, python just stores the memory reference or pointer itself and the actual value can be elsewhere in the memory.
Think of it like this:

````python
"""
List
+-----+-----+-----+
|  *  |  *  |  *  |
+-----+-----+-----+
   |      |      |
   v      v      v
  10    "hi"    3.14

"""
````

## 2. Does python allocates memory upfront?

Yes, but Partly.

Python allocates the small amount of memory upfront, since the List can be extended or appended in future, Python deosn't commit the entire world's worth of memory upfront. For example:

````python
# creating the empty list
my_list = []
````

Python allocates the small amount of memory upfront, it allocates some extra space. Let's understand this by example:

```
Capacity = 4
size = 2
[1][2][][]
```
When it becomes full
[1][2][3][4]

python allocates some more. Important point here is that Python doesn't allocate the memory with each operation, instead it allocates a larger block (say capacity 8), copies the references, and continues.

⚠️ The above example is just for illustration, Python doesn't exactly uses such logic, it can overallocate but on the basis of actual implementation (CPython)

## 3. Is a Python list memory efficient?

Yes - But it depends on what you mean.
Efficient because:
- The append operation is O(1)
- Random indexing is O(1).
- Resizing happens infrequently because of overallocation.

Not efficient because:
- Each element stored in list is a pointer object not the value itself.
  for example:

  ```
  my_list = [1,2,3]
  ```
  It doesn't store 4-byte array like C Array, instead it stores:

  ```
  my_list
    +-----+-----+-----+
    |  *  |  *  |  *  |
    +-----+-----+-----+
    |      |      |
    v      v      v
    PyInt   PyInt   PyInt
  ```
  Each integer is a full Python object (PyInt*) which contains the metadata as well.

## 4. Are Python lists like arrays in C?

| Feature                | C Array  | Python List                |
| ---------------------- | -------- | -------------------------- |
| Homogeneous            | Yes      | No                         |
| Stores values directly | Yes      | No (stores references)     |
| Contiguous memory      | Yes      | Contiguous references only |
| Dynamic size           | No       | Yes                        |
| Memory overhead        | Very low | Higher                     |
