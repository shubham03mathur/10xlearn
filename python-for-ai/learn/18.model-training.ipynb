# Model Training

Machine Learning is the process of teaching a computer to recognize patterns in data. This is done through a process called **Training**.

There are two main types of machine learning:
- **Supervised Learning**: The model learns from labeled data (e.g., predicting house prices based on previous sales).
- **Unsupervised Learning**: The model finds hidden patterns in unlabeled data (e.g., grouping customers into different segments based on buying behavior).

### Regression vs Classification
In Supervised Learning, we typically choose between two types of tasks:
1. **Regression**: Predicting a continuous number (e.g., Price, Temperature, Age).
2. **Classification**: Predicting a category (e.g., Spam vs Not Spam, Cat vs Dog).

### Training a Simple Regression Model
In this example, we will use **Scikit-Learn** to train a simple Linear Regression model to predict a value based on a single input feature.

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Sample data: X = Hours Studied, y = Score
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([10, 20, 30, 40, 50])

model = LinearRegression()
model.fit(X, y)

print(f"Coefficients: {model.coef_[0]}")
print(f"Intercept: {model.intercept_}")

# Make a prediction for 6 hours of study
prediction = model.predict([[6]])
print(f"Prediction for 6 hours: {prediction[0]}")
```

### Understanding the Training Process
When we call `model.fit()`, the computer is adjusting internal parameters (weights) to minimize the error between the predicted output and the actual output.

In more complex models like Neural Networks, this is done iteratively over multiple **Epochs** (passes through the data) and **Batches** (subsets of data).
