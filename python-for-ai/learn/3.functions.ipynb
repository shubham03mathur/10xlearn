# Function

### Defining function:

The keyword `def` is used to define the function in python followed by a function name (and its argument(s)) and body. Unlike other programming languages such as JavaScript, Python function doesn't have curley braces, instead the statements which forms the function body is defined in the next line and that should be indented.

> The first statement of the function body can be a string litrel, which is called docstring and its very important for 3rd party tools, specially in context of AI agents. I will discuss More about it later on.

Eveytime a function executes, python interpreter creates a new symbol table, which keep records of all variables inside the function, i.e. all variable assignments in a function stores the value in this new symbol table.

### What happens when python executes the function?
When python executes the function, it first check the reference for the variable in the local symbol table, if it doesn't find it there, then it looks in enclosing function's symbol table, If still it doens't find it then it moves to global symbol table and then at last in the built-in names. Thus, the global variable and the variables of enclosing functions cannot be assigned value directly inside a function (unless, for global variables, named in a global statement, or, for variables of enclosing functions, named in a nonlocal statement). Although, it can be referenced.

The actual paramenters (arguments) of a function are introduced into local symbol table of the called function, when it's called; thus, arguments are passed using call by value (where the value is always an object reference, not the value of the object).

Example:
Here is the function example in python - Writing a Fibonacci series to an arbitrary boundary:

```python
def fib(n: int) -> None:
    """ Print a fibonacci series less than n"""
    a, b = 0, 1
    while a < n:
        print(a, end=" ")
        a, b = b, a + b
    
fib(2000)
```

A function definition associates the function name with the function object/body in the current symbol table. The interpreter recognizes the object pointed to by that name as a user-defined function. That means we can assign the function to another variable, and that new variable will reference the same object and can be executed the same way (that's what first-class objects are, similar to JavaScript!).

```python
fib
```

```python
f = fib
f(100)
```

### Functional Arguments (or parameters)

Just like many other programming languages, Python also supports multiple type of arguments, such as positional, keyword based or default argument(s).

The most useful form is to create a function with one or more default arguments, this way the function can still be called with fewer arguments without fail. for example:


```python
def hey_xarvis(prompt, retries=4, reminder='Yes or no only!'):
    while True:
        reply = input(prompt)
        if reply in {'y', 'ye', 'yes', 'yea', 'yeah'}:
            return True
        if reply in {'n', 'no', 'nada', 'NO!', 'nope'}:
            return False
        retries = retries - 1
        if retries < 0:
            raise ValueError('invalid user response')
        print(reminder)
```

```python
hey_xarvis('Can you fly F-22 raptor?')
```

<div class="alert alert-block alert-warning"><b>Important:</b> The default value is evaluated only once, this is important because if the default value is set to any mutable object such as list, dictionary or class instances. For example the following function will keep accumulating the arguments passed to it on a subsequent calls</div>

```python
def demo(n, l=[]):
    l.append(n)
    return l

print(demo(1))
print(demo(2))
print(demo(3))
```

To fix this issue, we can use the following approach:

```python
def demo(n, l=None):
    if l is None:
        l = []
    l.append(n)
    return l

print(demo(1))
print(demo(2))
print(demo(3))
```
