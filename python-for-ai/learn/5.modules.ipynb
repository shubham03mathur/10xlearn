# Python Modules & Packages

Learn how Python modules organize code, create reusable packages, and manage imports effectively.

## What is a Module?

A **module** is a single file containing Python definitions and statements. When you run or import a file, it becomes a module.

```python
# Create your own module
def greet(name):
    return f"Hello, {name}!"

def calculate_average(numbers):
    if not numbers:
        return 0
    return sum(numbers) / len(numbers)
```

## Importing Modules

```python
# Save the above code as mymodule.py
# Then import it:

import mymodule

# Use its functions
print(mymodule.greet("World"))
print(mymodule.calculate_average([10, 20, 30]))
```

## Importing Specific Functions and Aliasing

```python
# Only import specific items
from mymodule import greet, calculate_average

print(greet("User"))

# Use aliases
from mymodule import calculate_average as calc_avg
print(calc_avg([5, 10, 15]))
```

## Package Structure

A **package** is a directory containing multiple modules. To make it a package, add an `__init__.py` file (Python 3.3+ can omit this but we'll include it).

```python
# Directory structure:
# mypackage/
# ├── __init__.py           # Makes it a package
# ├── util.py               # Module 1
# ├── math_ops.py          # Module 2
# └── string_ops.py        # Module 3

# In __init__.py you can expose public API:
from .util import *  # Selective imports in package
```

## Relative vs Absolute Imports

| Import Type | From Where | Example |
|--------------|-------------|----------|
| Absolute | Anywhere in project | `from mypackage.util import helper` |
| Relative | Inside package | `from .util import helper` |
| Parent-level | `..` prefix | `from ..shared import common` |

```python
# From within mypackage/math_ops.py:
from .util import process_data as pre_process

# From main.py (outside):
from mypackage.math_ops import run_analysis
```

## Import Order Matters

Python imports modules and executes their code. Be careful to avoid circular dependencies and runtime order issues.

```python
# GOOD: Import after defining classes that need imports
class Model:
    def __init__(self):
        # Import here to avoid circular deps
        from sklearn.preprocessing import StandardScaler
        self.scaler = StandardScaler()

# AVOID: This can cause issues:
# module_a.py has: from module_b import func
# module_b.py has: from module_a import class_x   # Circular!
```

## Best Practices

1. **Organize by feature, not type**
2. **Export specific functions, not everything with `from . import *`**
3. **Use `__all__` to declare public API explicitly**
4. **Keep private helpers with leading underscore prefix**
5. **Avoid deep nesting - limit package depth to 2-3 levels max**

```python
# Declare public API
__all__ = ["greet", "calculate_average", "process_data"]

# Private helper functions (don't import with leading _)
def __internal_helper():  # Start with underscore
    """Not part of public API"""
    pass
```

## When to Create a Module vs Use Standard Library

| Situation | Recommendation |
|-----------|---------------|
| Reuse code across files | Create module |
| Share with others | Package it properly |
| Simple script, one-purpose | Just be a script file |
| Common functionality? | Consider standard library first |
