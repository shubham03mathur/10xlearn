# Deployment Basics

Deployment is the process of taking a trained machine learning model and making it accessible for real-world use. 

When a model is 'deployed', it means other users or applications can send it data and receive predictions. This is how models like ChatGPT, Netflix recommendations, or FaceID work.

### Common Deployment Methods

1. **Web APIs**: The model is wrapped in a web server. A user sends a request (e.g., via an HTTP request), and the server returns a JSON prediction.
2. **Embedded Systems**: The model runs directly on a device (e.g., a camera detecting a face locally).
3. **Batch Processing**: The model runs on a large group of data at a scheduled time (e.g., calculating credit scores for all bank customers once a night).

### Popular Tools for Deployment

Several frameworks make it easy to deploy models in Python:
- **Flask**: A lightweight web framework for creating APIs.
- **FastAPI**: A modern, high-performance web framework for building APIs with Python.
- **Streamlit**: The fastest way to build and share simple ML web apps (great for data science demos).

### MLOps (Machine Learning Operations)

Deployment is just the beginning. **MLOps** is the practice of managing the entire lifecycle of a model, which includes:
- **Monitoring**: Tracking the model's performance in production.
- **Data Drift**: Detecting when the real-world data starts to look different than the training data.
- **Retraining**: Automatically updating the model with new data to keep it accurate.
