# Python Package

As a popular open source developement project, Python has cast library of packages or pre-buil modules which are battle tested and ready to use. so rather than re-inventing the wheel, we can just install/download these packages from official repository - PyPI (Python Package Index).

Install your first package.

```  pip install <package-name>  ```

for example: ``` pip install requests ```

#### Key Terms:
- pip is a python package manager and is preffered package installer which ships with python itself.
- `venv` or virtual environment is a isolated runtime environment which allows use and installation of 
   packages for a specific app rather then installing them globally, This is important because in Linux environment updating global packages or even switching to other python version can cause issues.
- The Python Package Index (PyPI) is a public repository of open source licensed packages made available
  for use by other python users.

> Changed in version 3.5: The use of venv is now recommended for creating virtual environments.

````python
# Install requests package using 
# `pip install requests` OR `uv pip install requests` (if you are using uv venv)

import requests

res = requests.get("https://api.github.com")
print(res.status_code)
````
